## 1. โหลดเครื่องมือและข้อมูล

```python
import pandas as pd
import matplotlib.pyplot as plt

BASE = "https://raw.githubusercontent.com/Pattharaporn-Aon/StatBUU2569-workshop/main/data/"   # ที่อยู่ของข้อมูลบน GitHub

df = pd.read_csv(BASE + "02_water_level_daily.csv")
df.head()   # ดู 5 แถวแรก
```

## 2. สรุปค่าสถิติพื้นฐาน

```python
print("จำนวนแถว, คอลัมน์ =", df.shape)
df.describe().round(1)
```

## 3. ฮิสโทแกรม: ระดับน้ำกระจายตัวแบบไหน?

```python
plt.hist(df["water_level_m"], bins=20, color="#5cb8e6", edgecolor="black")
plt.xlabel("Water level (m)")
plt.ylabel("จำนวนวัน (Frequency)")
plt.title("Histogram of daily river water level")
plt.show()

print("ค่าเฉลี่ย =", round(df["water_level_m"].mean(), 2), " มัธยฐาน =", round(df["water_level_m"].median(), 2))
```

**คำถาม:** ค่าเฉลี่ยมากกว่าหรือน้อยกว่ามัธยฐาน? การกระจายเบ้ไปทางไหน?

## 4. Boxplot รายเดือน: เดือนไหนน้ำขึ้นสูง?

```python
df.boxplot(column="water_level_m", by="month", grid=False, figsize=(9, 4))
plt.xlabel("Month")
plt.ylabel("Water level (m)")
plt.suptitle("")
plt.title("River water level by month")
plt.show()
```

## 5. นักสืบข้อมูล: หาความผิดปกติใน dirty data

```python
dirty = pd.read_csv(BASE + "03_water_level_daily_dirty.csv")

print("ค่าว่างในแต่ละคอลัมน์:")
print(dirty.isna().sum())
print("\nจำนวนแถวซ้ำ:", dirty.duplicated().sum())
print("\nระดับน้ำที่ติดลบหรือสูงผิดปกติ (> 10 เมตร):")
dirty[(dirty["water_level_m"] < 0) | (dirty["water_level_m"] > 10)]
```

## 6. สหสัมพันธ์ ≠ เหตุและผล

```python
ice = pd.read_csv(BASE + "04_icecream_drowning.csv")
print(ice[["icecream_sales", "drowning_cases", "avg_temp_c"]].corr().round(2))

plt.scatter(ice["icecream_sales"], ice["drowning_cases"], color="#f9a23b", s=80)
plt.xlabel("Ice cream sales")
plt.ylabel("Drowning cases")
plt.show()
```

**คำถาม:** สรุปว่าไอศกรีมทำให้คนจมน้ำได้ไหม? ตัวแปรไหนคือตัวแปรแทรกซ้อน?

## 7. Anscombe's Quartet: ตัวเลขเหมือนกัน แต่กราฟต่างกัน

```python
ans = pd.read_csv(BASE + "05_anscombe.csv")
print(ans.groupby("dataset")[["x", "y"]].mean().round(2))

fig, axes = plt.subplots(2, 2, figsize=(8, 6), sharex=True, sharey=True)
colors = ["#ffde59", "#5ce1e6", "#ff8fab", "#7ed957"]
for ax, (name, g), c in zip(axes.flat, ans.groupby("dataset"), colors):
    ax.scatter(g["x"], g["y"], color=c, edgecolor="black", s=60)
    ax.set_title(f"Dataset {name}   r = {g['x'].corr(g['y']):.2f}")
plt.tight_layout()
plt.show()
```
